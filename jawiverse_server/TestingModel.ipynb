### 📊 Evaluation Metrics for Transliteration Model

This section explains the evaluation metrics commonly used to assess **Jawi ↔ Rumi transliteration** quality.  
We use four main metrics: **BLEU**, **ChrF**, **CER**, and **WER**.

---

##### 🧮 1. BLEU (Bilingual Evaluation Understudy)

**Purpose:**  
Measures how closely the predicted output matches the reference at the **word level**.

**How it works:**
- Compares overlapping **n-grams** (1-gram, 2-gram, etc.) between the model output and the reference text.
- Produces a score between `0` and `1`.

**Interpretation:**
| BLEU Score | Meaning |
|-------------|----------|
| 1.0 | Perfect match |
| 0.7–1.0 | Excellent |
| 0.4–0.7 | Moderate |
| <0.4 | Weak |

**Notes:**
- BLEU is case- and token-sensitive.
- Works well for **translation**, but can be harsh for **transliteration**, since one wrong letter can drop the score.

---

##### 🔡 2. ChrF (Character F-score)

**Purpose:**  
Evaluates character-level accuracy — ideal for **transliteration tasks**.

**How it works:**
- Measures **character n-gram** overlaps instead of words.
- Computes the **F1-score** (precision + recall) for characters.
- Returns a score between `0` and `100`.

**Interpretation:**
| ChrF Score | Meaning |
|-------------|----------|
| 100 | Perfect match |
| >90 | Excellent |
| 70–90 | Good |
| <70 | Many character errors |

**Notes:**
- Better suited than BLEU for transliteration.
- More robust against spacing and word boundary issues.

---

##### 🧠 3. CER (Character Error Rate)
---
**Purpose:**  
Measures literal character accuracy.

**How it works:**
- Based on **edit distance** (Levenshtein distance).
- Calculates how many characters must be **inserted, deleted, or substituted** to match the reference.
- Formula:  
  \[
  CER = \frac{\text{Edits}}{\text{Total Reference Characters}}
  \]

**Interpretation:**
| CER Score | Meaning |
|------------|----------|
| 0.0 | Perfect |
| 0.1 | 10% of characters wrong |
| 0.3+ | Weak performance |

**Notes:**
- Every single character counts (including punctuation, spacing, and casing).
- Very precise for **transliteration fidelity**.

### 2. Compute evaluation metrics

#### Evaluation Metrics Function

```python
def compute_metrics(predictions, references):
    """
    Compute BLEU, ChrF, CER, and WER scores for text predictions.
    Returns values rounded to 4 decimal places.
    """

    # Convert to string list
    predictions = [str(p) for p in predictions]
    references = [str(r) for r in references]

    # Load metrics
    bleu_metric = evaluate.load("bleu")
    chrf_metric = evaluate.load("chrf")

    # Compute BLEU
    bleu_score = bleu_metric.compute(
        predictions=predictions,
        references=[[r] for r in references]  # BLEU expects nested lists
    )["bleu"]

    # Compute ChrF
    chrf_score = chrf_metric.compute(
        predictions=predictions,
        references=[[r] for r in references]
    )["score"]

    # Compute CER & WER
    cer_score = cer(references, predictions)
    wer_score = wer(references, predictions)

    # Return rounded results
    return {
        "BLEU": round(bleu_score, 4),
        "ChrF": round(chrf_score, 4),
        "CER": round(cer_score, 4),
        "WER": round(wer_score, 4)
    }
```

### 3. Inference

#### Transliteration Generation Function
Example: Jawi → Rumi

```python
from unsloth.chat_templates import get_chat_template
from transformers import TextStreamer
import torch

def transliterate_generate(model, tokenizer, prompt: str,
                           direction: str = "jawi2rumi",
                           chat_template="llama-3.1",
                           max_new_tokens=128,
                           temperature=0.1,
                           min_p=0.1):
    """
    Generate transliteration (Jawi↔Rumi) using Unsloth fine-tuned model.

    Args:
        model: Loaded FastLanguageModel
        tokenizer: Corresponding tokenizer
        prompt (str): Input text (either Jawi or Rumi)
        direction (str): Either 'jawi2rumi' or 'rumi2jawi'
        chat_template (str): Chat template used during training
        max_new_tokens (int): Max tokens to generate
        temperature (float): Sampling temperature
        min_p (float): Minimum nucleus sampling probability

    Returns:
        str: Generated transliteration
    """

    # Ensure correct chat template
    tokenizer = get_chat_template(tokenizer, chat_template=chat_template)

    # Optional: live text streamer
    text_streamer = TextStreamer(tokenizer, skip_prompt=True, skip_special_tokens=True)

    # ✅ Set tag based on direction
    if direction not in ["jawi2rumi", "rumi2jawi"]:
        raise ValueError("Invalid direction. Use 'jawi2rumi' or 'rumi2jawi'.")

    messages = [
        {"role": "user", "content": f"<{direction}> {prompt}"},
    ]

    # Tokenize
    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt",
    ).to(model.device)

    # Attention mask
    attention_mask = inputs.ne(tokenizer.pad_token_id).long()

    # Generate
    with torch.no_grad():
        outputs = model.generate(
            input_ids=inputs,
            attention_mask=attention_mask,
            streamer=text_streamer,
            max_new_tokens=max_new_tokens,
            temperature=temperature,
            min_p=min_p,
        )

    generated = tokenizer.decode(outputs[0], skip_special_tokens=True)

    # 🧹 Clean up Unsloth / Llama chat template junk
    if "assistant" in generated:
        generated = generated.split("assistant", 1)[-1]
    elif "<|assistant|>" in generated:
        generated = generated.split("<|assistant|>", 1)[-1]

    return generated.strip()
```

##### a. Manual examples

###### Evaluate manual examples function

```python
def evaluate_manual_examples(model, tokenizer, examples, direction="jawi2rumi", preview_rows=5):
    """
    Evaluate manual Jawi↔Rumi transliteration examples with metrics.

    Args:
        model: Fine-tuned Unsloth model
        tokenizer: Tokenizer for the model
        examples: List of dicts [{"Input": ..., "Actual": ...}, ...]
        direction: "jawi2rumi" or "rumi2jawi"
        preview_rows: Number of example rows to display in the output

    Returns:
        comparison_df: DataFrame with Input, Predicted, and Actual
        metrics: dict with BLEU, ChrF, CER, WER
    """
    manual_df = pd.DataFrame(examples)
    results = []

    # --- Generate predictions ---
    print(f"🪶 Transliteration Evaluation: {direction}")
    for i, row in tqdm(manual_df.iterrows(), total=len(manual_df), desc=f"Generating {direction}"):
        input_text = str(row["Input"])
        actual_output = str(row["Actual"])

        predicted_output = transliterate_generate(model, tokenizer, prompt=input_text, direction=direction)

        results.append({
            "No": i + 1,
            "Input": input_text,
            "Predicted": predicted_output.strip(),
            "Actual": actual_output.strip()
        })

    comparison_df = pd.DataFrame(results)
    pd.set_option('display.max_colwidth', 200)

    # --- Compute evaluation metrics ---
    metrics = compute_metrics(
        predictions=comparison_df["Predicted"].tolist(),
        references=comparison_df["Actual"].tolist()
    )

    # --- Display results ---
    print("\n📘 Example Predictions:")
    display(comparison_df.head(preview_rows))

    print("\n📊 Evaluation Metrics:")
    for k, v in metrics.items():
        print(f"{k}: {v:.4f}")

    return comparison_df, metrics
```

```python
examples = [
    {"Input": "كمدين ممولڠكنڽ كڤد أصلڽ",
     "Actual": "Kemudian memulangkannya kepada asalnya"},
    # {"Input": "چينتا سجاتي اداله سوات ڤركارا يڠ مرجوع كڤد حقيقة باطن",
    #  "Actual": "Cinta sejati adalah suatu perkara yang merujuk kepada hakikat batin"}
]

comparison_df, metrics = evaluate_manual_examples(model, tokenizer, examples, direction="jawi2rumi")
```

##### b. N test samples (random)


###### Unified Evaluation Function (with random_state + progress info)

```python
def evaluate_transliteration(
    model,
    tokenizer,
    file_path: str,
    direction: str = "jawi2rumi",
    sheet_name: str = None,      # None = evaluate all sheets
    n_samples: int = 10,         # number of samples per sheet
    preview_rows: int = 5,       # how many rows to display
    random_state: int = 42,      # seed for reproducibility / random sampling
    save_results: bool = False   # save CSV summaries
):
    """
    Evaluate Jawi↔Rumi transliteration model on one or all Excel sheets.

    Args:
        model: fine-tuned Unsloth model
        tokenizer: corresponding tokenizer
        file_path: path to Excel dataset
        direction: "jawi2rumi" or "rumi2jawi"
        sheet_name: specific sheet to evaluate, or None for all
        n_samples: number of random samples per sheet
        preview_rows: number of sample predictions to display
        random_state: random seed for sampling reproducibility
        save_results: whether to save CSV outputs
    """

    # Define input/output column names
    input_col = "Jawi" if direction == "jawi2rumi" else "Rumi"
    target_col = "Rumi" if direction == "jawi2rumi" else "Jawi"

    results_summary = []

    # --- Determine which sheets to process ---
    if sheet_name:
        sheet_list = [sheet_name]
    else:
        sheet_list = pd.ExcelFile(file_path).sheet_names
        print(f"📄 Found {len(sheet_list)} sheets: {sheet_list}\n")

    # --- Iterate over sheets ---
    for idx, sheet in enumerate(sheet_list, start=1):
        print(f"\n🟢 [{idx}/{len(sheet_list)}] Now processing sheet: '{sheet}' ...")

        df = pd.read_excel(file_path, sheet_name=sheet)

        # Skip if missing columns
        if input_col not in df.columns or target_col not in df.columns:
            print(f"⚠️ Skipping '{sheet}' (missing {input_col}/{target_col})")
            continue

        # Sample data with random_state
        sample_df = df.sample(min(n_samples, len(df)), random_state=random_state).reset_index(drop=True)

        # Generate predictions
        rows = []
        for i, row in tqdm(sample_df.iterrows(), total=len(sample_df), desc=f"🔄 Generating {direction} [{sheet}]"):
            input_text = str(row[input_col])
            actual_output = str(row[target_col])
            predicted_output = transliterate_generate(model, tokenizer, prompt=input_text, direction=direction)
            rows.append({
                "Sheet": sheet,
                "No": i + 1,
                "Input": input_text,
                "Predicted": predicted_output.strip(),
                "Actual": actual_output.strip()
            })

        # Create comparison DataFrame
        comparison_df = pd.DataFrame(rows)

        # Compute metrics
        metrics = compute_metrics(
            predictions=comparison_df["Predicted"].tolist(),
            references=comparison_df["Actual"].tolist()
        )

        results_summary.append({
            "Sheet": sheet,
            "Samples": len(sample_df),
            **metrics,
        })

        # Optional preview
        print(f"\n📘 Sample Predictions from sheet '{sheet}':")
        display(comparison_df.head(preview_rows))

        # Save per-sheet comparison if enabled
        if save_results:
            out_path = f"results_{sheet}_{direction}.csv"
            comparison_df.to_csv(out_path, index=False)
            print(f"💾 Saved detailed results to: {out_path}")

        print("-" * 80)  # Separator between sheets

    # --- Create overall summary ---
    summary_df = pd.DataFrame(results_summary)
    print("\n📊 Overall Evaluation Summary:")
    display(summary_df)

    if save_results:
        summary_path = f"evaluation_summary_{direction}.csv"
        summary_df.to_csv(summary_path, index=False)
        print(f"✅ Saved metrics summary to '{summary_path}'")

    return summary_df
```
